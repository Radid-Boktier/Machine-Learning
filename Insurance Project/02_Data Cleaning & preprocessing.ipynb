```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings('ignore')
```

### Load the dataset

```python
df = pd.read_csv('insurance.csv')
```

### Data Cleaning and preprocessing

```python
df_cleaned = df.copy()
```

```python
df_cleaned.head()
```

```python
df_cleaned.shape
```

## Remove Duplicate Records
Duplicate records can unnecessarily increase dataset size and processing time. Removing them also prevents repeated observations from having disproportionate influence during model training.

```python
df_cleaned.drop_duplicates(inplace = True)
```

```python
df_cleaned.shape
```

## Check for Missing Values
Check each feature for missing values because they may cause errors or unreliable results during model training. This helps determine whether imputation or another missing-value handling strategy is required.

```python
df_cleaned.isnull().sum()
```

```python
df_cleaned.dtypes
```

```python
df_cleaned['sex'].value_counts()
```

### Encode Categorical Features
Inspect the unique values in the sex and smoker features and apply `Label Encoding` to convert their binary categories into numerical values. This preserves the categorical information in a model-compatible format.

```python
df_cleaned['sex'] = df_cleaned['sex'].map({"male" : 0, "female" : 1})
```

```python
df_cleaned.head()
```

```python
df_cleaned['smoker'].value_counts()
```

```python
df_cleaned['smoker'] = df_cleaned['smoker'].map({"no" : 0, "yes" : 1})
```

```python
df_cleaned.head()
```

### Rename Encoded Features

```python
df_cleaned.rename(columns={
    'sex' : 'is_female',
    'smoker': 'is_smoker'
                            }, inplace = True)
```

```python
df_cleaned.head()
```

```python
df['region'].value_counts()
```

### One-Hot Encode region
Apply one-hot encoding because the regional categories have no natural numerical ordering. This represents each category independently without introducing an artificial relationship between regions.

```python
df_cleaned = pd.get_dummies(df_cleaned, columns = ['region'])
```

```python
df_cleaned.head()
```

### Convert Encoded Features to Integers
Convert Boolean indicator features into integer values represented by 0 and 1. This provides a consistent numerical representation for downstream preprocessing and model training.

```python
df_cleaned = df_cleaned.astype(int)
```

```python
df_cleaned
```

```python
df_cleaned.to_csv("cleaned_insurance.csv", index=False)
```
